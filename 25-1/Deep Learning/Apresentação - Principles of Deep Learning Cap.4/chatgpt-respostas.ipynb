A fórmula que você apresentou faz parte do formalismo de integrais funcionais (ou integrais de caminho) em teoria de campos, estatística ou física matemática, geralmente no contexto de teorias perturbativas de matrizes aleatórias ou integrais gaussianas com perturbações. Vamos destrinchar a lógica de como se chega da equação (4.46) à equação (4.47).

---

### **1. Interpretação da fórmula (4.46)**

Você tem uma expectativa (esperança) sobre uma função arbitrária $F$, denotada por:

$$
\mathbb{E}[F(z)] = \frac{\int [\prod_{i,\alpha} dz_{i;\alpha}] \, e^{-S(z)} F(z)}{\int [\prod_{i,\alpha} dz_{i;\alpha}] \, e^{-S(z)}}
$$

Com uma ação $S(z)$ do tipo:

$$
S(z) = \frac{1}{2} \sum_{i,\alpha,\beta} z_{i;\alpha} (g^{-1})_{\alpha\beta} z_{i;\beta} - \frac{1}{8} \sum_{\beta_1 \ldots \beta_4} v^{(\beta_1 \beta_2)(\beta_3 \beta_4)} \sum_{j_1,j_2} z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4}
$$

Essa é uma perturbação quartica sobre uma teoria gaussiana (o primeiro termo da ação é puramente gaussiano).

---

### **2. Expansão perturbativa (até primeira ordem em $v$)**

A expectativa é expandida em série de potências no potencial $v$, levando à fórmula:

$$
\langle\langle F(z) \rangle\rangle_v = \langle\langle F(z) \rangle\rangle_g + \frac{1}{8} \sum_{\beta_1 \ldots \beta_4} v^{(\beta_1 \beta_2)(\beta_3 \beta_4)} \sum_{j_1, j_2} \left[ \langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} F(z) \rangle\rangle_g - \langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} \rangle\rangle_g \langle\langle F(z) \rangle\rangle_g \right] + O(v^2)
$$

Ou seja, aplicamos a **expansão de cumulantes (ou de cumulantes conectados)** da expectativa de $F$ com a perturbação quartica.

---

### **3. Aplicando isso ao caso específico de $F(z) = z_{i_1;\alpha_1} z_{i_2;\alpha_2}$**

O próximo passo é aplicar a fórmula anterior para:

$$
F(z) = z_{i_1;\alpha_1} z_{i_2;\alpha_2}
$$

Então, usando a equação (4.46), temos:

$$
\mathbb{E}[z_{i_1;\alpha_1} z_{i_2;\alpha_2}] = \langle\langle z_{i_1;\alpha_1} z_{i_2;\alpha_2} \rangle\rangle_g + \frac{1}{8} \sum_{\beta_1 \ldots \beta_4} v^{(\beta_1 \beta_2)(\beta_3 \beta_4)} \sum_{j_1,j_2} \Bigg[
\langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} z_{i_1;\alpha_1} z_{i_2;\alpha_2} \rangle\rangle_g
- \langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} \rangle\rangle_g \langle\langle z_{i_1;\alpha_1} z_{i_2;\alpha_2} \rangle\rangle_g
\Bigg] + O(v^2)
$$

---

### **4. Cálculo com integrais gaussianas (usando Wick)**

Nas integrais gaussianas, usamos o **teorema de Wick**, que nos permite calcular médias de produtos de variáveis gaussianas como somas de produtos de suas contrações. O valor médio gaussiano de $z_{i;\alpha} z_{j;\beta}$ é:

$$
\langle\langle z_{i;\alpha} z_{j;\beta} \rangle\rangle_g = \delta_{ij} g_{\alpha\beta}
$$

Assim, os termos de sexta ordem como:

$$
\langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} z_{i_1;\alpha_1} z_{i_2;\alpha_2} \rangle\rangle_g
$$

podem ser decompostos em somas de produtos de pares usando Wick, o que gera diversos termos do tipo:

* $g_{\beta_1\beta_2} g_{\beta_3\beta_4} g_{\alpha_1\alpha_2}$
* $g_{\beta_1\alpha_1} g_{\beta_2\alpha_2} g_{\beta_3\beta_4}$
* etc., respeitando as condições de índice $\delta_{ij}$

---

### **5. Resultado final: equação (4.47)**

A média perturbada fica:

$$
\mathbb{E}[z_{i_1;\alpha_1} z_{i_2;\alpha_2}] = \delta_{i_1 i_2} \left[
g_{\alpha_1\alpha_2} + \frac{1}{2} \sum_{\beta_1,\ldots,\beta_4} v^{(\beta_1 \beta_2)(\beta_3 \beta_4)} \left( n g_{\alpha_1\beta_1} g_{\alpha_2\beta_2} g_{\beta_3\beta_4} + 2n g_{\alpha_1\beta_1} g_{\alpha_2\beta_3} g_{\beta_2\beta_4} \right)
\right] + O(v^2)
$$

Esse resultado usa:

* $\sum_{j_1, j_2} \delta_{j_1 i_1} \delta_{j_2 i_2} \to n$ (pela média sobre $j$)
* Todas as contrações possíveis entre os 6 campos, separando os termos conectados (que contribuem para a correção de ordem $v$).

---

### **Resumo do processo:**

1. **Partimos da ação perturbada**, com termo gaussiano + quartico.
2. **Expandimos a expectativa perturbada em série em $v$**, usando cumulantes.
3. **Aplicamos essa fórmula ao caso específico** $F = z_{i_1;\alpha_1} z_{i_2;\alpha_2}$.
4. **Usamos Wick para calcular as médias gaussianas** de 6 campos.
5. **Somamos todos os termos**, respeitando simetrias e contrações, até obter o resultado final.

Se quiser, posso detalhar explicitamente o cálculo dos termos de Wick que contribuem para (4.47). Deseja isso?


Perfeito! Vamos então **detalhar a aplicação do teorema de Wick** para o termo:

$$
\left\langle\left\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} z_{i_1;\alpha_1} z_{i_2;\alpha_2} \right\rangle\right\rangle_g
$$

com o objetivo de obter a correção de ordem $v$ para:

$$
\mathbb{E}[z_{i_1;\alpha_1} z_{i_2;\alpha_2}]
$$

Lembrando que:

* A média gaussiana de qualquer número **ímpar** de variáveis é **zero**.
* A média de um produto de **seis variáveis gaussianas com média zero** é dada pela soma de todos os **produtos de pares possíveis** (contratações) — existem **15 termos possíveis**.

Como estamos interessados apenas em termos que **contribuem a ordem $v$**, devemos considerar apenas os **termos conectados** com:

$$
F(z) = z_{i_1;\alpha_1} z_{i_2;\alpha_2}
$$

---

## ✅ **Passo 1: Expressão geral**

Vamos expandir:

$$
\langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} z_{i_1;\alpha_1} z_{i_2;\alpha_2} \rangle\rangle_g
$$

pela regra de Wick. Queremos **apenas os termos com contrações que envolvam os índices $\alpha_1, \alpha_2$** (isto é, termos que de fato contribuem com $F$).

---

## ✅ **Passo 2: Contrações relevantes**

Vamos considerar $i_1 = i_2 = i$ (pois a correção só existe quando $\delta_{i_1 i_2} = 1$).

A média:

$$
\langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} z_{i;\alpha_1} z_{i;\alpha_2} \rangle\rangle_g
$$

tem termos com contrações do tipo:

### 🔹 Tipo A:

* $z_{j_1;\beta_1} \contract z_{i;\alpha_1}$,
* $z_{j_1;\beta_2} \contract z_{i;\alpha_2}$,
* $z_{j_2;\beta_3} \contract z_{j_2;\beta_4}$

Esse dá:

$$
\delta_{j_1 i} \delta_{j_1 i} \delta_{j_2 j_2} \, g_{\beta_1 \alpha_1} g_{\beta_2 \alpha_2} g_{\beta_3 \beta_4}
$$

Somando sobre $j_1, j_2$, temos:

$$
\sum_{j_1, j_2} \delta_{j_1 i} \delta_{j_1 i} = \sum_{j_2} \delta_{j_2 j_2} = n
$$

Contribuição total do tipo A:

$$
n \, g_{\beta_1 \alpha_1} g_{\beta_2 \alpha_2} g_{\beta_3 \beta_4}
$$

### 🔹 Tipo B: troca $\alpha_2$ com $\beta_3$

* $z_{j_1;\beta_1} \contract z_{i;\alpha_1}$
* $z_{j_2;\beta_3} \contract z_{i;\alpha_2}$
* $z_{j_1;\beta_2} \contract z_{j_2;\beta_4}$

Contribuição:

$$
\delta_{j_1 i} \delta_{j_2 i} \, g_{\beta_1 \alpha_1} g_{\beta_3 \alpha_2} g_{\beta_2 \beta_4}
$$

Somando sobre $j_1, j_2$:

$$
\sum_{j_1, j_2} \delta_{j_1 i} \delta_{j_2 i} = n
$$

Contribuição total do tipo B:

$$
n \, g_{\beta_1 \alpha_1} g_{\beta_3 \alpha_2} g_{\beta_2 \beta_4}
$$

### 🔹 Tipo C: troca $\alpha_1 \leftrightarrow \alpha_2$ do anterior

* $z_{j_1;\beta_1} \contract z_{i;\alpha_2}$,
* $z_{j_2;\beta_3} \contract z_{i;\alpha_1}$,
* $z_{j_1;\beta_2} \contract z_{j_2;\beta_4}$

Resultado:

$$
n \, g_{\beta_1 \alpha_2} g_{\beta_3 \alpha_1} g_{\beta_2 \beta_4}
$$

---

## ✅ **Passo 3: Soma dos tipos relevantes**

Agrupando os três tipos relevantes (e os simétricos), a contribuição total à correção de primeira ordem de:

$$
\mathbb{E}[z_{i;\alpha_1} z_{i;\alpha_2}]
$$

fica:

$$
\delta_{i_1 i_2} \left[
g_{\alpha_1 \alpha_2} + \frac{1}{2} \sum_{\beta_1 \ldots \beta_4} v^{(\beta_1 \beta_2)(\beta_3 \beta_4)} \left( n g_{\alpha_1 \beta_1} g_{\alpha_2 \beta_2} g_{\beta_3 \beta_4} + 2n g_{\alpha_1 \beta_1} g_{\alpha_2 \beta_3} g_{\beta_2 \beta_4} \right)
\right] + O(v^2)
$$

A **fator $\frac{1}{2}$** aparece ao reorganizar os 3 termos simétricos e levar em conta a simetria de $v^{(\beta_1 \beta_2)(\beta_3 \beta_4)}$, assumida como simétrica sob trocas de pares.

---

### ✅ Conclusão

Portanto, o termo:

$$
\mathbb{E}[z_{i_1;\alpha_1} z_{i_2;\alpha_2}]
$$

resulta em:

$$
\delta_{i_1 i_2} \left[
g_{\alpha_1 \alpha_2} + \frac{1}{2} \sum_{\beta_1 \ldots \beta_4} v^{(\beta_1 \beta_2)(\beta_3 \beta_4)} \left( n g_{\alpha_1 \beta_1} g_{\alpha_2 \beta_2} g_{\beta_3 \beta_4} + 2n g_{\alpha_1 \beta_1} g_{\alpha_2 \beta_3} g_{\beta_2 \beta_4} \right)
\right] + O(v^2)
$$

exatamente como dado na equação **(4.47)**.

Ótima pergunta — e muito importante para entender completamente como a expansão funciona.

Quando aplicamos o **teorema de Wick** para calcular médias gaussianas de produtos de variáveis (como $\langle\langle zzzzzz \rangle\rangle_g$), há **vários termos possíveis** de contração. No caso de 6 variáveis, existem **15 termos de pareamento**.

Contudo, **nem todos os 15 termos contribuem na correção da expectativa** de $\mathbb{E}[z_{i_1;\alpha_1} z_{i_2;\alpha_2}]$. Vamos entender **por quê** e **o que acontece com os outros termos**.

---

### ✅ Contexto: expansão de cumulantes

Você tem a fórmula (repetindo aqui de forma compacta):

$$
\mathbb{E}[F(z)] = \langle\langle F(z) \rangle\rangle_g + \frac{1}{8} \sum_{\text{índices}} v \sum_{j_1, j_2} \left[
\langle\langle Q(z) F(z) \rangle\rangle_g - \langle\langle Q(z) \rangle\rangle_g \langle\langle F(z) \rangle\rangle_g
\right] + \mathcal{O}(v^2)
$$

onde $Q(z) = z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4}$ é o termo quartico da perturbação.

---

### ✅ Termos restantes de Wick: o que acontece com eles?

Vamos dividir os **15 termos de Wick** em **3 categorias**:

---

### 🔹 **Categoria 1: Termos desconectados**

Estes são termos onde as variáveis $z_{i_1;\alpha_1}$ e $z_{i_2;\alpha_2}$ **não estão conectadas** com os $z_{j;\beta}$, por exemplo:

$$
\langle\langle z_{j_1;\beta_1} z_{j_1;\beta_2} z_{j_2;\beta_3} z_{j_2;\beta_4} \rangle\rangle_g \cdot \langle\langle z_{i_1;\alpha_1} z_{i_2;\alpha_2} \rangle\rangle_g
$$

Esse é **exatamente o termo subtraído** no termo de correção:

$$
\langle\langle Q(z) \rangle\rangle_g \cdot \langle\langle F(z) \rangle\rangle_g
$$

Portanto, **esses termos se cancelam** — **são eliminados** na diferença. Só sobram os termos **conectados**, ou seja, os que misturam variáveis de $Q(z)$ e de $F(z)$.

---

### 🔹 **Categoria 2: Termos onde $i_1 \ne i_2$**

Se $i_1 \ne i_2$, então qualquer contração que dependa de $\delta_{i_1 i_2}$ será zero.

Lembre que:

$$
\langle\langle z_{i_1;\alpha_1} z_{i_2;\alpha_2} \rangle\rangle_g = \delta_{i_1 i_2} g_{\alpha_1 \alpha_2}
$$

Logo, se $i_1 \ne i_2$, o termo de ordem zero é zero e os de ordem $v$ também são zero **porque as contrações resultam em produtos com $\delta_{i_1 i_2} = 0$**. Portanto, esses termos **também não contribuem**.

---

### 🔹 **Categoria 3: Termos com contrações “misturadas” mas redundantes**

Alguns termos fazem contrações de $z_{i_1;\alpha_1}$ com $z_{i_2;\alpha_2}$, e os outros quatro entre si, por exemplo:

* $z_{i_1;\alpha_1} \contract z_{i_2;\alpha_2}$,
* $z_{j_1;\beta_1} \contract z_{j_1;\beta_2}$,
* $z_{j_2;\beta_3} \contract z_{j_2;\beta_4}$

Esses também aparecem **tanto no termo completo $\langle\langle QF \rangle\rangle_g$** quanto no produto fatorado $\langle\langle Q \rangle\rangle_g \langle\langle F \rangle\rangle_g$\*\*. Então eles também **se cancelam**.

---

## ✅ Conclusão

👉 **Somente os termos realmente conectados** entre o termo quartico $Q(z)$ e $F(z) = z_{i;\alpha_1} z_{i;\alpha_2}$ **contribuem para a correção de primeira ordem**.

🔻 Os outros termos:

* **Se cancelam** porque aparecem nos dois termos da subtração;
* **São nulos** por conta de $\delta_{i_1 i_2} = 0$;
* **Não estão conectados**, então não afetam os cumulantes de ordem superior.
